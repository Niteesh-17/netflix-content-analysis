## 6. Movie Duration Analysis

In this section, we select movie records and convert their duration values into numerical minutes. This makes it easier to perform calculations and analyze movie lengths.

```python

movie = df[df["type"] == "Movie"].copy()

movie["duration_minutes"] = movie["duration"].str.replace(
    " min", "", regex=False
)

movie["duration_minutes"] = pd.to_numeric(
    movie["duration_minutes"], errors="coerce"
)

longest_movie = movie.sort_values(
    by="duration_minutes", ascending=False
)

longest_movie[
    ["title", "release_year", "duration", "duration_minutes"]
].head(10)
```

## 7. Grouping and Aggregation

The `groupby()` function is used to calculate the number of titles released in each year. We also identify the year with the highest number of releases in the dataset.

### Data Visualization Using Matplotlib

Matplotlib is used to represent the analysis visually. The charts help us understand yearly release trends, the distribution of movies and TV shows, content ratings, and other patterns in the dataset.

This bar chart shows the number of Netflix titles grouped by their release year.  
It helps identify the years in which the largest number of titles in the dataset were released.

```python
titles_released=df.groupby("release_year").size().sort_index()
print(titles_released)
```

```python
maxx_titles=titles_released.idxmax()
print(maxx_titles)
print(titles_released.max())
```

```python
plt.figure(figsize=(12,5))
titles_released.plot(kind="bar",color="orange")
plt.title("Title released per year")
plt.xlabel("Release year ")
plt.ylabel("Number of Titles")
plt.xticks(
    ticks=range(0, len(titles_released), 5),
    labels=titles_released.index[::5],
    rotation=45
)
plt.tight_layout()
plt.show()
```

This bar chart compares the number of Movies and TV Shows in the Netflix dataset.

It helps understand the distribution of content types available in the dataset.

```python
content_count=df["type"].value_counts()
plt.figure(figsize=(7, 5))
content_count.plot(kind="bar",color=["royalblue","red"])
plt.title("Movies vs TV Shows")
plt.xlabel("Content types")
plt.ylabel("number of Viewers")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

This bar chart displays the ten most frequent content ratings in the Netflix dataset.

It helps identify the ratings most commonly assigned to the available titles.

```python
top_rating=df["rating"].value_counts().head(10)
plt.figure(figsize=(10,6))
top_rating.plot(kind="bar",color=["red","blue","green","orange","royalblue"])
plt.title("Top 10 Content Ratings on Netflix")
plt.xlabel("Ratings")
plt.ylabel("Number of Titles")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

This line chart shows the number of movies released in each year.

It helps visualize changes in movie releases over time.

```python
movie= df[df["type"]=="Movie"]
movie_released_peryear=movie.groupby("release_year").size().sort_index()
plt.figure(figsize=(10, 6))
movie_released_peryear.plot(kind="line", color="blue")
plt.title("Movies Released per Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Movies")
plt.grid(True)
plt.tight_layout()
plt.show()

```

This histogram illustrates the distribution of movie durations in minutes.

It helps identify common duration ranges and understand how movie lengths vary across the dataset.

```python
type_mov=df[df["type"]=="Movie"].copy()
type_mov["duration_minutes"]=type_mov["duration"].str.replace(" min", "", regex=False)
type_mov["duration_minutes"] = pd.to_numeric(
type_mov["duration_minutes"], errors="coerce"
)
durations = type_mov["duration_minutes"].dropna()
plt.figure(figsize=(10, 6))
plt.hist(durations, bins=15, edgecolor="black", color="skyblue")
plt.title("Distribution of Netflix Movie Durations")
plt.xlabel("Movie Duration (Minutes)")
plt.ylabel("Number of Movies")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()
plt.show()



```

This horizontal bar chart displays the ten most frequent country entries in the dataset.

This pie chart shows the percentage share of Movies and TV Shows in the dataset.

It provides a quick visual comparison of the proportions of the two content types.

```python
content_count = df["type"].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    content_count,
    labels=content_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribution of Movies and TV Shows on Netflix")
plt.tight_layout()
plt.show()
```

This scatter plot examines the relationship between movie release years and movie durations.

It helps reveal patterns in movie lengths over time and highlights potentially unusual durations.

```python

type_mov = df[df["type"] == "Movie"].copy()
type_mov["duration_minutes"] = type_mov["duration"].str.replace(
    " min", "", regex=False
)

type_mov["duration_minutes"] = pd.to_numeric(
    type_mov["duration_minutes"], errors="coerce"
)

type_mov["release_year"] = pd.to_numeric(
    type_mov["release_year"], errors="coerce"
)

scatter_data = type_mov.dropna(
    subset=["release_year", "duration_minutes"]
)


plt.figure(figsize=(10, 6))

plt.scatter(
    scatter_data["release_year"],
    scatter_data["duration_minutes"],
    alpha=0.5
)


plt.title("Movie Duration vs. Release Year")
plt.xlabel("Release Year")
plt.ylabel("Movie Duration (Minutes)")

plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
